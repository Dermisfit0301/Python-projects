**Channel Analysis**

In this assignment, our objective is to analyze various channels and find out the most effective channel to optimize ad spend based on metrics like high client conversion rates, revenue to spend ratio etc. Based on these insights we will make suggestions to our client.

```python
#Loading the dataset
import pandas as pd
df=pd.read_excel('/content/sample_data/Traffic data.xlsx')
df.head()


```

```python
#examining the dataset
df.info()
```

```python
# Function to convert 'hh:mm:ss' format into total seconds
def time_to_seconds(time_str):

    if isinstance(time_str, str):
        h, m, s = map(int, time_str.split(':'))
    else:

        h = time_str.hour
        m = time_str.minute
        s = time_str.second
    return h * 3600 + m * 60 + s
```

```python
# Removing the dollar sign and commas from the 'Revenue' column and convert to float
df['Revenue'] = df['Revenue'].replace({'\$': '', ',': ''}, regex=True).astype(float)
df['Revenue'].head()
```

```python

# Applying the conversion function to the 'Avg. Session Duration' column
df['Avg. Session Duration'] = df['Avg. Session Duration'].apply(time_to_seconds)
df['Avg. Session Duration'].head()
```

```python
#new user percentage
df['New User Percentage'] = df['New Users']/df['Users']*100
df['New User Percentage']=df['New User Percentage'].round(2)
df['New User Percentage']=df['New User Percentage'].astype(int)
df['New User Percentage'].head()
```

```python
#getting the number of customers
df['Customers'] = df['Users']*df['Ecommerce Conversion Rate']
df['Customers']=df['Customers'].round()
df['Customers']=df['Customers'].astype(int)
df['Customers'].head()
```

```python
#getting revenue per customer
df['Revenue per Customer'] = df['Revenue']/df['Customers']
df['Revenue per Customer']=df['Revenue per Customer'].round(2)
df['Revenue per Customer'].head()
```

```python
#getting revenue per transaction
df['Revenue per Transaction'] = df['Revenue']/df['Transactions']
df['Revenue per Transaction']=df['Revenue per Transaction'].round(2)
df['Revenue per Transaction'].head()
```

```python
#Getting sessions per user
df['Sessions per User'] = df['Sessions']/df['Users']
df['Sessions per User']=df['Sessions per User'].round(2)
df['Sessions per User'].head()
```

```python
#getting transaction volumne
df['Transaction Volume'] = df['Transactions']/df['Sessions']
df['Transaction Volume']=df['Transaction Volume'].round(3)
df['Transaction Volume'].head()
```

```python
#examining the new dataset
df.head()
```

```python
# Strip any leading/trailing spaces from column names
df.columns = df.columns.str.strip()
```

```python
# The best Pay-per-click (PPC) Advertising Sources by Revenue



ppc_df = df[df['Medium'].str.contains('cpc', case=False)]
revenue_by_ppc = ppc_df.groupby('Source').sum()['Revenue'].sort_values(ascending=False)
revenue_by_ppc
```

```python

# Importing necessary libraries

import matplotlib.pyplot as plt
import seaborn as sns
```

```python

# Plotting to see revenue by ppc source
plt.figure(figsize=(10, 6))
sns.barplot(x=revenue_by_ppc.index, y=revenue_by_ppc.values, palette="Reds_d")
plt.title('Viewing PPC Advertising Sources by Revenue')
plt.ylabel('Total Revenue')
plt.xlabel('Source')
plt.show()
```

```python
#Segregating variables
engagement_columns = ['New User Percentage', 'Sessions per User', 'Pages / Session', 'Avg. Session Duration', 'Bounce Rate']
revenue_columns = ['Revenue', 'Revenue per Customer', 'Revenue per Transaction', 'Ecommerce Conversion Rate', 'Transactions','Customers']
```

```python
#Analysing engagement metrics for source


def plot_engagement_metrics(df):
    engagement_data = df.groupby('Source')[engagement_columns].mean()

    # Plot multiple metrics
    fig, axes = plt.subplots(3, 2, figsize=(16, 12))

    sns.barplot(x=engagement_data.index, y=engagement_data['New User Percentage'], ax=axes[0, 1])
    axes[0, 1].set_title('New User Percentage per Source')
    axes[0, 1].set_xticklabels(engagement_data.index, rotation=90)

    sns.barplot(x=engagement_data.index, y=engagement_data['Sessions per User'], ax=axes[1, 0])
    axes[1, 0].set_title('Sessions per User per Source')
    axes[1, 0].set_xticklabels(engagement_data.index, rotation=90)

    sns.barplot(x=engagement_data.index, y=engagement_data['Pages / Session'], ax=axes[1, 1])
    axes[1, 1].set_title('Pages per Session per Source')
    axes[1, 1].set_xticklabels(engagement_data.index, rotation=90)

    sns.barplot(x=engagement_data.index, y=engagement_data['Avg. Session Duration'], ax=axes[2, 0])
    axes[2, 0].set_title('Avg. Session Duration (seconds) per Source')
    axes[2, 0].set_xticklabels(engagement_data.index, rotation=90)

    sns.barplot(x=engagement_data.index, y=engagement_data['Bounce Rate'], ax=axes[2, 1])
    axes[2, 1].set_title('Bounce Rate per Source')
    axes[2, 1].set_xticklabels(engagement_data.index, rotation=90)

    plt.tight_layout()
    plt.show()

plot_engagement_metrics(df)
```

```python
# Plotting revenue metrics for each Source
def plot_revenue_metrics(df):
    revenue_data = df.groupby('Source')[revenue_columns].mean()

    # Plot multiple metrics
    fig, axes = plt.subplots(3, 2, figsize=(16, 12))

    sns.barplot(x=revenue_data.index, y=revenue_data['Revenue'], ax=axes[0, 0])
    axes[0, 0].set_title('Average Revenue per Source')
    axes[0, 0].set_xticklabels(revenue_data.index, rotation=90)

    sns.barplot(x=revenue_data.index, y=revenue_data['Revenue per Customer'], ax=axes[0, 1])
    axes[0, 1].set_title('Revenue per Customer per Source')
    axes[0, 1].set_xticklabels(revenue_data.index, rotation=90)

    sns.barplot(x=revenue_data.index, y=revenue_data['Revenue per Transaction'], ax=axes[1, 0])
    axes[1, 0].set_title('Revenue per Transaction per Source')
    axes[1, 0].set_xticklabels(revenue_data.index, rotation=90)

    sns.barplot(x=revenue_data.index, y=revenue_data['Ecommerce Conversion Rate'], ax=axes[1, 1])
    axes[1, 1].set_title('Ecommerce Conversion Rate per Source')
    axes[1, 1].set_xticklabels(revenue_data.index, rotation=90)

    sns.barplot(x=revenue_data.index, y=revenue_data['Transactions'], ax=axes[2, 0])
    axes[2, 0].set_title('Average Transactions per Source')
    axes[2, 0].set_xticklabels(revenue_data.index, rotation=90)

    sns.barplot(x=revenue_data.index, y=revenue_data['Customers'], ax=axes[2, 0])
    axes[2, 0].set_title('Average Transactions per Source')
    axes[2, 0].set_xticklabels(revenue_data.index, rotation=90)


    plt.tight_layout()
    plt.show()

# Use the function
plot_revenue_metrics(df)
```
